# BAB 2 PEMBAHASAN

## 2.1 Dasar Teori

Operator adalah simbol atau kata kunci yang digunakan untuk melakukan operasi terhadap satu atau lebih nilai dalam program. Python memiliki beberapa jenis operator dengan fungsi yang berbeda. Pada praktikum ini digunakan tujuh kelompok operator, yaitu operator aritmatika, perbandingan, logika, penugasan, keanggotaan, identitas, dan bitwise.

### 2.1.1 Operator Aritmatika

Operator aritmatika digunakan untuk melakukan operasi matematika pada nilai numerik. Operator yang umum digunakan dalam Python meliputi:

| Operator | Fungsi | Contoh |
|---|---|---|
| `+` | Penjumlahan | `5 + 2` |
| `-` | Pengurangan | `5 - 2` |
| `*` | Perkalian | `5 * 2` |
| `/` | Pembagian | `5 / 2` |
| `//` | Pembagian bulat | `5 // 2` |
| `%` | Sisa bagi | `5 % 2` |
| `**` | Pangkat | `5 ** 2` |

Operator aritmatika digunakan untuk menghasilkan nilai baru berdasarkan operasi matematika yang dilakukan terhadap operand.

### 2.1.2 Operator Perbandingan

Operator perbandingan digunakan untuk membandingkan dua nilai. Hasil operasi perbandingan berupa nilai Boolean, yaitu `True` atau `False`.

Operator yang digunakan antara lain:

| Operator | Fungsi |
|---|---|
| `==` | Sama dengan |
| `!=` | Tidak sama dengan |
| `>` | Lebih besar |
| `<` | Lebih kecil |
| `>=` | Lebih besar atau sama dengan |
| `<=` | Lebih kecil atau sama dengan |

Operator perbandingan dapat digunakan untuk menentukan apakah suatu kondisi terpenuhi.

### 2.1.3 Operator Logika

Operator logika digunakan untuk menggabungkan atau membalik kondisi Boolean.

- `and` menghasilkan `True` jika seluruh kondisi bernilai `True`.
- `or` menghasilkan `True` jika minimal salah satu kondisi bernilai `True`.
- `not` membalik nilai Boolean suatu kondisi.

Operator logika banyak digunakan dalam proses validasi dan pengambilan keputusan pada program.

### 2.1.4 Operator Penugasan (Assignment)

Operator penugasan digunakan untuk memberikan atau memperbarui nilai suatu variabel. Operator dasar penugasan adalah `=`. Python juga menyediakan augmented assignment seperti `+=`, `-=`, `*=`, `/=`, dan operator sejenisnya.

Contoh:

```python
nilai_2024 = 10
nilai_2024 += 5
```

Setelah operasi tersebut, nilai variabel menjadi 15.

### 2.1.5 Operator Keanggotaan

Operator keanggotaan digunakan untuk memeriksa apakah suatu nilai terdapat dalam suatu objek atau koleksi.

- `in` menghasilkan `True` jika nilai ditemukan.
- `not in` menghasilkan `True` jika nilai tidak ditemukan.

Contoh:

```python
promo_2024 = "HEMAT10"
daftar_promo_2024 = ["HEMAT10", "HEMAT20"]

promo_2024 in daftar_promo_2024
```

Hasilnya adalah `True` karena `HEMAT10` terdapat dalam daftar promo.

### 2.1.6 Operator Identitas

Operator identitas digunakan untuk membandingkan identitas dua objek.

- `is` digunakan untuk memeriksa apakah dua variabel menunjuk objek yang sama.
- `is not` digunakan untuk memeriksa apakah dua variabel menunjuk objek yang berbeda.

Operator `is` berbeda dengan `==`. Operator `==` membandingkan nilai atau isi objek, sedangkan `is` membandingkan identitas objek.

### 2.1.7 Operator Bitwise

Operator bitwise melakukan operasi terhadap representasi biner suatu bilangan. Pada praktikum ini digunakan:

| Operator | Fungsi |
|---|---|
| `&` | AND bitwise |
| `|` | OR bitwise |
| `^` | XOR bitwise |
| `~` | NOT bitwise |
| `<<` | Shift kiri |
| `>>` | Shift kanan |

Operator bitwise dapat digunakan untuk merepresentasikan beberapa kondisi menggunakan bit tertentu dan memeriksa status tersebut secara efisien.

## 2.2 Kode Program

Bagian ini berisi program-program yang digunakan untuk mempraktikkan setiap jenis operator. Kode program ditempatkan pada **code cell**, sedangkan penjelasan setiap program ditempatkan pada **markdown cell** setelah kode.

### 2.2.1 `aritmatika_2611532024.py`

````python
# Buat file dengan nama aritmatika_2611532024.py
# Buat program untuk operator aritmatika dalam Python
# Nama variabel ditambah 4 digit nim terakhir contoh: angka1_1234
# Program ini menggunakan fungsi input()
# Nilai yang dimasukkan akan dikonversi menjadi tipe data integer

angka1_2024 = int(input("Input angka-1: "))
angka2_2024 = int(input("Input angka-2: "))

# Penjumlahan
hasil_2024 = angka1_2024 + angka2_2024
print("\nOperator Penjumlahan")
print("Hasil =", hasil_2024)

# Pengurangan
hasil_2024 = angka1_2024 - angka2_2024
print("\nOperator Pengurangan")
print("Hasil =", hasil_2024)

# Perkalian
hasil_2024 = angka1_2024 * angka2_2024
print("\nOperator Perkalian")
print("Hasil =", hasil_2024)

# Pembagian, pembagian bulat, dan sisa bagi
if angka2_2024 != 0:
    hasil_2024 = angka1_2024 / angka2_2024
    print("\nOperator Pembagian")
    print("Hasil =", hasil_2024)

    hasil_2024 = angka1_2024 // angka2_2024
    print("\nOperator Pembagian Bulat")
    print("Hasil =", hasil_2024)

    hasil_2024 = angka1_2024 % angka2_2024
    print("\nOperator Sisa Bagi")
    print("Hasil =", hasil_2024)
else:
    print("Angka kedua tidak boleh bernilai 0.")

# Pangkat
hasil_2024 = angka1_2024 ** angka2_2024
print("\nOperator Pangkat")
print("Hasil =", hasil_2024)
````

**Penjelasan Program**

Program meminta pengguna memasukkan dua bilangan melalui fungsi `input()`. Kedua input dikonversi menjadi tipe data integer menggunakan `int()`. Selanjutnya program menerapkan operator penjumlahan (`+`), pengurangan (`-`), perkalian (`*`), pembagian (`/`), pembagian bulat (`//`), sisa bagi (`%`), dan pangkat (`**`).

Sebelum melakukan pembagian, program memeriksa apakah `angka2_2024` tidak sama dengan nol menggunakan operator perbandingan `!=`. Pemeriksaan tersebut mencegah operasi pembagian dengan nol. Setiap hasil operasi disimpan pada variabel `hasil_2024` dan kemudian ditampilkan.

Kode ini juga menunjukkan bahwa nama variabel mengikuti aturan empat digit terakhir NIM, yaitu `_2024`. Program tersebut merupakan penerapan langsung operator aritmatika sesuai kode yang diberikan.

### 2.2.2 `perbandingan_2611532024.py`

````python
# Buat file dengan nama perbandingan_2611532024.py
# Nama variabel ditambah 4 digit nim terakhir contoh: angka1_1234
# Program ini menggunakan fungsi input()
# Nilai yang dimasukkan akan dikonversi menjadi tipe data integer
# Program operator perbandingan dalam Python

angka1_2024 = int(input("Input angka-1: "))
angka2_2024 = int(input("Input angka-2: "))

# Lebih besar dari
hasil_2024 = angka1_2024 > angka2_2024
print("\nOperator lebih besar dari")
print("angka1 > angka2 =", hasil_2024)

# Lebih kecil dari
hasil_2024 = angka1_2024 < angka2_2024
print("\nOperator lebih kecil dari")
print("angka1 < angka2 =", hasil_2024)

# Lebih besar dari atau sama dengan
hasil_2024 = angka1_2024 >= angka2_2024
print("\nOperator lebih besar dari atau sama dengan")
print("angka1 >= angka2 =", hasil_2024)

# Lebih kecil dari atau sama dengan
hasil_2024 = angka1_2024 <= angka2_2024
print("\nOperator lebih kecil dari atau sama dengan")
print("angka1 <= angka2 =", hasil_2024)

# Sama dengan
hasil_2024 = angka1_2024 == angka2_2024
print("\nOperator sama dengan")
print("angka1 == angka2 =", hasil_2024)

# Tidak sama dengan
hasil_2024 = angka1_2024 != angka2_2024
print("\nOperator tidak sama dengan")
print("angka1 != angka2 =", hasil_2024)

# Tambahan: perbandingan berantai dalam Python
hasil_2024 = 0 < angka1_2024 < 100
print("\nPerbandingan berantai")
print("0 < angka1 < 100 =", hasil_2024)

hasil_2024 = 0 < angka2_2024 < 100
print("0 < angka2 < 100 =", hasil_2024)
````

**Penjelasan Program**

Program meminta dua bilangan integer dari pengguna menggunakan `input()` dan `int()`. Selanjutnya kedua bilangan dibandingkan menggunakan operator `>`, `<`, `>=`, `<=`, `==`, dan `!=`. Setiap hasil perbandingan disimpan dalam `hasil_2024` dan ditampilkan sebagai nilai Boolean `True` atau `False`.

Program juga menggunakan perbandingan berantai seperti `0 < angka1_2024 < 100` untuk memeriksa apakah suatu angka berada dalam rentang 0 sampai 100. Dengan demikian, program menunjukkan penggunaan operator perbandingan secara langsung pada input pengguna.

### 2.2.3 `logika_2611532024.py`

````python
# Buat file dengan nama logika_2611532024.py
# Nama variabel ditambah 4 digit nim terakhir contoh: a1_1234
# Program ini menggunakan fungsi input()
# Program operator logika dalam Python

# Memasukkan nilai boolean
# Input tidak peka terhadap huruf besar dan kecil
a1_2024 = input("Input nilai boolean-1 (true/false): ").strip().lower() == "true"
a2_2024 = input("Input nilai boolean-2 (true/false): ").strip().lower() == "true"

print("\nA1 =", a1_2024)
print("A2 =", a2_2024)

# Konjungsi: bernilai True jika keduanya True
hasil_2024 = a1_2024 and a2_2024
print("\nKonjungsi (AND)")
print("A1 and A2 =", hasil_2024)

# Disjungsi: bernilai True jika salah satunya True
hasil_2024 = a1_2024 or a2_2024
print("\nDisjungsi (OR)")
print("A1 or A2 =", hasil_2024)

# Negasi A1: membalik nilai A1
hasil_2024 = not a1_2024
print("\nNegasi A1 (NOT)")
print("not A1 =", hasil_2024)

# Negasi A2: membalik nilai A2
hasil_2024 = not a2_2024
print("\nNegasi A2 (NOT)")
print("not A2 =", hasil_2024)

# XOR: bernilai True jika kedua nilai berbeda
hasil_2024 = a1_2024 != a2_2024
print("\nDisjungsi Eksklusif (XOR)")
print("A1 XOR A2 =", hasil_2024)
````

**Penjelasan Program**

Program meminta dua nilai Boolean dari pengguna. Input diubah menjadi huruf kecil menggunakan `lower()` dan kemudian dibandingkan dengan string `"true"` sehingga menghasilkan nilai Boolean.

Selanjutnya program menerapkan operator logika `and`, `or`, dan `not`. Operator `and` digunakan untuk konjungsi, `or` untuk disjungsi, dan `not` untuk membalik nilai Boolean. Program juga menunjukkan XOR melalui perbandingan `!=`, yang menghasilkan `True` ketika kedua nilai Boolean berbeda.

### 2.2.4 `assignment_2611532024.py`

````python
# Buat file dengan nama assignment_2611532024.py
# Nama variabel ditambah 4 digit nim terakhir contoh: angka1_1234
# Program ini menggunakan fungsi input()
# Nilai yang dimasukkan akan dikonversi menjadi tipe data integer
# Program operator assignment dalam Python

angka1_2024 = int(input("Input angka-1: "))
angka2_2024 = int(input("Input angka-2: "))

print("\nNilai awal angka1 =", angka1_2024)
print("Nilai angka2 =", angka2_2024)

# Assignment biasa
hasil_2024 = angka1_2024
print("\nAssignment biasa (=)")
print("Hasil =", hasil_2024)

# Assignment penambahan
hasil_2024 = angka1_2024
hasil_2024 += angka2_2024
print("\nAssignment penambahan (+=)")
print("Hasil =", hasil_2024)

# Assignment pengurangan
hasil_2024 = angka1_2024
hasil_2024 -= angka2_2024
print("\nAssignment pengurangan (-=)")
print("Hasil =", hasil_2024)

# Assignment perkalian
hasil_2024 = angka1_2024
hasil_2024 *= angka2_2024
print("\nAssignment perkalian (*=)")
print("Hasil =", hasil_2024)

# Assignment pembagian, pembagian bulat, dan sisa bagi
if angka2_2024 != 0:
    hasil_2024 = angka1_2024
    hasil_2024 /= angka2_2024
    print("\nAssignment pembagian (/=)")
    print("Hasil =", hasil_2024)
    # Operator tambahan
    hasil_2024 = angka1_2024
    hasil_2024 //= angka2_2024
    print("\nAssignment pembagian bulat (//=)")
    print("Hasil =", hasil_2024)
    hasil_2024 = angka1_2024
    hasil_2024 %= angka2_2024
    print("\nAssignment sisa bagi (%=)")
    print("Hasil =", hasil_2024)
else:
    print("\nPembagian tidak dapat dilakukan.")
    print("Angka kedua tidak boleh bernilai 0.")

# Operator tambahan: assignment perpangkatan
hasil_2024 = angka1_2024
hasil_2024 **= angka2_2024
print("\nAssignment perpangkatan (**=)")
print("Hasil =", hasil_2024)
````

**Penjelasan Program**

Program meminta dua bilangan integer. Nilai pertama kemudian digunakan sebagai dasar untuk berbagai operasi assignment. Assignment biasa menggunakan `=`, sedangkan augmented assignment yang digunakan meliputi `+=`, `-=`, `*=`, `/=`, `//=`, `%=`, dan `**=`.

Sebelum operasi pembagian dilakukan, program memeriksa apakah angka kedua tidak bernilai nol menggunakan `!=`. Setiap operasi memperbarui nilai `hasil_2024`, sehingga program memperlihatkan perbedaan antara assignment biasa dan augmented assignment.

### 2.2.5 `lainnya_2611532024.py`

````python
# Buat file dengan nama lainnya_2611532024.py
# Nama variabel ditambah 4 digit nim terakhir contoh: angka1_1234
# Program ini menggunakan fungsi input()
# Program operator keanggotaan dan  identitas

print("========================================")
print("1. OPERATOR KEANGGOTAAN")
print("========================================")

# Input beberapa data yang dipisahkan dengan koma
input_data_2024 = input("Masukkan beberapa angka, pisahkan dengan koma: ")

# Mengubah input menjadi list integer
data_2024 = [int(angka.strip()) for angka in input_data_2024.split(",")]

nilai_dicari_2024 = int(input("Masukkan angka yang ingin dicari: "))

# Operator in
hasil_2024 = nilai_dicari_2024 in data_2024
print("\nOperator keanggotaan IN")
print(nilai_dicari_2024, "in", data_2024, "=", hasil_2024)

# Operator not in
hasil_2024 = nilai_dicari_2024 not in data_2024
print("\nOperator keanggotaan NOT IN")
print(nilai_dicari_2024, "not in", data_2024, "=", hasil_2024)


print("\n========================================")
print("2. OPERATOR IDENTITAS")
print("========================================")

# objek1 menggunakan list dari input pengguna
objek1_2024 = data_2024

# objek2 merujuk pada objek yang sama dengan objek1
objek2_2024 = objek1_2024

# objek3 memiliki isi sama, tetapi merupakan objek baru
objek3_2024 = data_2024.copy()

print("objek1 =", objek1_2024)
print("objek2 =", objek2_2024)
print("objek3 =", objek3_2024)

# Operator is
hasil = objek1_2024 is objek2_2024
print("\nOperator identitas IS")
print("objek1 is objek2 =", hasil_2024)

# Operator is not
hasil_2024 = objek1_2024 is not objek3_2024
print("\nOperator identitas IS NOT")
print("objek1 is not objek3 =", hasil_2024)

# Membandingkan identitas dan nilai
print("\nPerbandingan identitas dan nilai")
print("objek1 is objek3 =", objek1_2024 is objek3_2024)
print("objek1 == objek3 =", objek1_2024 == objek3_2024)
````

**Penjelasan Program**

Program terdiri atas dua bagian, yaitu operator keanggotaan dan operator identitas. Pada bagian keanggotaan, pengguna memasukkan beberapa angka yang dipisahkan dengan koma. Data tersebut kemudian diubah menjadi list integer. Operator `in` digunakan untuk memeriksa apakah nilai yang dicari terdapat dalam list, sedangkan `not in` memeriksa apakah nilai tersebut tidak terdapat dalam list.

Pada bagian identitas, `objek2_2024` dibuat untuk merujuk objek yang sama dengan `objek1_2024`, sedangkan `objek3_2024` dibuat sebagai salinan baru menggunakan `copy()`. Oleh karena itu, `objek1_2024 is objek2_2024` menunjukkan identitas objek yang sama, sedangkan `objek1_2024 is not objek3_2024` menunjukkan objek yang berbeda. Program juga membandingkan `is` dengan `==` untuk menunjukkan perbedaan antara identitas dan nilai.

### 2.2.6 `bitwise_2611532024.py`

````python
# Buat file dengan nama bitwise_2611532024.py
# Nama variabel ditambah 4 digit nim terakhir contoh: angka1_1234
# Program ini menggunakan fungsi input()

print("\n========================================")
print("3. OPERATOR BITWISE")
print("========================================")

angka1_2024 = int(input("Masukkan angka bitwise-1: "))
angka2_2024 = int(input("Masukkan angka bitwise-2: "))

print("\nAngka dalam bentuk desimal dan biner")
print("angka1 =", angka1_2024, "| biner =", bin(angka1_2024))
print("angka2 =", angka2_2024, "| biner =", bin(angka2_2024))

# Bitwise AND
hasil_2024 = angka1_2024 & angka2_2024
print("\nBitwise AND (&)")
print(angka1_2024, "&", angka2_2024, "=", hasil_2024)
print("Biner hasil =", bin(hasil_2024))
print("Biner hasil (8 bit) =", format(hasil_2024, "08b"))

# Bitwise OR
hasil_2024 = angka1_2024 | angka2_2024
print("\nBitwise OR (|)")
print(angka1_2024, "|", angka2_2024, "=", hasil_2024)
print("Biner hasil =", bin(hasil_2024))
print("Biner hasil (8 bit) =", format(hasil_2024, "08b"))

# Bitwise XOR
hasil_2024 = angka1_2024 ^ angka2_2024
print("\nBitwise XOR (^)")
print(angka1_2024, "^", angka2_2024, "=", hasil_2024)
print("Biner hasil =", bin(hasil_2024))
print("Biner hasil (8 bit) =", format(hasil_2024, "08b"))

# Bitwise NOT
hasil_2024 = ~angka1_2024
print("\nBitwise NOT (~)")
print("~", angka1_2024, "=", hasil_2024)
print("Biner hasil =", bin(hasil_2024))
print("Biner hasil (8 bit) =", format(hasil_2024, "08b"))

# Bitwise geser kiri
jumlah_geser_2024 = int(input("\nMasukkan jumlah pergeseran bit: "))

hasil_2024 = angka1_2024 << jumlah_geser_2024
print("\nBitwise geser kiri (<<)")
print(angka1_2024, "<<", jumlah_geser_2024, "=", hasil_2024)
print("Biner hasil =", bin(hasil_2024))
print("Biner hasil (8 bit) =", format(hasil_2024, "08b"))

# Bitwise geser kanan
hasil_2024 = angka1_2024 >> jumlah_geser_2024
print("\nBitwise geser kanan (>>)")
print(angka1_2024, ">>", jumlah_geser_2024, "=", hasil_2024)
print("Biner hasil =", bin(hasil_2024))
print("Biner hasil (8 bit) =", format(hasil_2024, "08b"))
````

**Penjelasan Program**

Program meminta dua bilangan integer kemudian menampilkan bentuk desimal dan biner masing-masing. Selanjutnya digunakan operator bitwise AND (`&`), OR (`|`), XOR (`^`), dan NOT (`~`).

Program juga menggunakan operator geser kiri (`<<`) dan geser kanan (`>>`). Jumlah pergeseran dimasukkan oleh pengguna. Fungsi `bin()` digunakan untuk menampilkan bentuk biner, sedangkan `format(..., "08b")` digunakan untuk menampilkan hasil dalam format biner 8 bit. Program ini menunjukkan bagaimana operasi bitwise bekerja pada representasi biner suatu bilangan.

### Studi Kasus: Sistem Simulasi Transaksi dan Validasi Akses Toko

Program kasir sederhana digunakan untuk menghitung total transaksi pelanggan sekaligus melakukan pemeriksaan terhadap status pelanggan dan hak akses terhadap promo tertentu. Program menerima data transaksi dan informasi pelanggan, kemudian menggunakan berbagai jenis operator Python untuk menentukan total pembayaran, status diskon, kelayakan promo, serta hak akses pelanggan.

Seluruh proses dibuat dalam satu alur program yang utuh dengan memanfaatkan operator aritmatika, perbandingan, logika, penugasan, keanggotaan, identitas, dan bitwise.

### 2.3.1 `tugas3_2611532024.py`

````python
# ===SISTEM TRANSAKSI TOKO===
print("=== SISTEM TRANSAKSI TOKO ===")

# ===INPUT DATA PELANGGAN===
nama_2024 = input("\nMasukkan Nama Pelanggan : ")
status_2024 = input("Masukkan Status Pelanggan (member/nonmember) : ").lower()
belanja_2024 = float(input("Masukkan Total Belanja : "))
jumlah_2024 = int(input("Masukkan Jumlah Barang : "))
promo_2024 = input("Masukkan Kode Promo : ").upper()

# ===DATA PROMO===
daftar_promo_2024 = ["HEMAT10", "HEMAT20", "GRATISONGKIR"]

# ===OPERATOR PERBANDINGAN===
belanja_minimum_2024 = belanja_2024 >= 200000
jumlah_minimum_2024 = jumlah_2024 >= 3
status_member_2024 = status_2024 == "member"

# ===OPERATOR KEANGGOTAAN===
promo_tersedia_2024 = promo_2024 in daftar_promo_2024
promo_tidak_tersedia_2024 = promo_2024 not in daftar_promo_2024

# ===OPERATOR LOGIKA===

# AND: member dan memenuhi minimum belanja
diskon_member_2024 = status_member_2024 and belanja_minimum_2024

# AND: jumlah barang memenuhi syarat dan promo tersedia
promo_khusus_2024 = jumlah_minimum_2024 and promo_tersedia_2024

# OR: mendapatkan promo jika memenuhi salah satu kondisi
mendapatkan_promo_2024 = diskon_member_2024 or promo_khusus_2024

# NOT: mengecek apakah pelanggan bukan member
bukan_member_2024 = not status_member_2024

# ===OPERATOR ARITMATIKA===
if diskon_member_2024:
    persentase_diskon_2024 = 0.10
else:
    persentase_diskon_2024 = 0.00

# Perkalian (*)
diskon_2024 = belanja_2024 * persentase_diskon_2024

# Pengurangan (-)
total_pembayaran_2024 = belanja_2024 - diskon_2024

# Pembagian (/)
rata_rata_2024 = belanja_2024 / jumlah_2024

# Modulus (%)
sisa_2024 = belanja_2024 % 1000

# ===OPERATOR PENUGASAN===
poin_2024 = 0

# Augmented assignment (+=)
poin_2024 += jumlah_2024

# Augmented assignment (-=)
if bukan_member_2024:
    poin_2024 -= 1

# ===HAK AKSES PELANGGAN===
member_access_2024 = status_member_2024
promo_access_2024 = promo_tersedia_2024
free_shipping_2024 = promo_2024 == "GRATISONGKIR"

# Membentuk kode hak akses
kode_hak_akses_2024 = 0

if member_access_2024:
    kode_hak_akses_2024 += 1

if promo_access_2024:
    kode_hak_akses_2024 += 2

if free_shipping_2024:
    kode_hak_akses_2024 += 4

# ===OPERATOR IDENTITAS===

# Membuat dua objek dengan nilai yang sama
kode_1_2024 = ["HEMAT10"]
kode_2_2024 = ["HEMAT10"]

# kode_3 menunjuk objek yang sama dengan kode_1
kode_3_2024 = kode_1_2024

identitas_sama_2024 = kode_1_2024 is kode_3_2024
identitas_berbeda_2024 = kode_1_2024 is not kode_2_2024

# == membandingkan nilai
nilai_sama_2024 = kode_1_2024 == kode_2_2024

# ===OPERATOR BITWISE===

# Nilai bit:
# 0001 = Member
# 0010 = Belanja >= Rp200.000
# 0100 = Jumlah barang >= 3
# 1000 = Kode promo tersedia

bit_member_2024 = 0b0001
bit_belanja_2024 = 0b0010
bit_jumlah_2024 = 0b0100
bit_promo_2024 = 0b1000

# OR (|) untuk menggabungkan status
kode_status_2024 = 0

if status_member_2024:
    kode_status_2024 = kode_status_2024 | bit_member_2024

if belanja_minimum_2024:
    kode_status_2024 = kode_status_2024 | bit_belanja_2024

if jumlah_minimum_2024:
    kode_status_2024 = kode_status_2024 | bit_jumlah_2024

if promo_tersedia_2024:
    kode_status_2024 = kode_status_2024 | bit_promo_2024

# AND (&) untuk memeriksa status tertentu
cek_member_2024 = kode_status_2024 & bit_member_2024
cek_promo_2024 = kode_status_2024 & bit_promo_2024

# XOR (^) untuk membandingkan dua kode
kode_referensi_2024 = 0b1011
perbandingan_status_2024 = kode_status_2024 ^ kode_referensi_2024

# SHIFT (<<)
hasil_shift_2024 = kode_status_2024 << 1

# OUTPUT DATA TRANSAKSI
print("\n")
print("=== DATA TRANSAKSI ===")

print("Nama Pelanggan       :", nama_2024)
print("Status Pelanggan     :", status_2024)
print("Total Belanja        : Rp" + str(int(belanja_2024)))
print("Jumlah Barang        :", jumlah_2024)
print("Kode Promo           :", promo_2024)

# OUTPUT HASIL VALIDASI
print("\n")
print("=== HASIL VALIDASI ===")

print("Belanja >= Rp200000        :", belanja_minimum_2024)
print("Jumlah Barang >= 3         :", jumlah_minimum_2024)
print("Status Member              :", status_member_2024)
print("Kode Promo Tersedia        :", promo_tersedia_2024)
print("Mendapatkan Diskon         :", diskon_member_2024)
print("Mendapatkan Promo          :", mendapatkan_promo_2024)

# OUTPUT HASIL PERHITUNGAN
print("\n")
print("=== HASIL PERHITUNGAN ===")

print("Diskon                  : Rp" + str(int(diskon_2024)))
print("Total Pembayaran        : Rp" + str(int(total_pembayaran_2024)))
print("Rata-rata Harga Barang  : Rp" + str(int(rata_rata_2024)))

# OUTPUT HAK AKSES PELANGGAN
print("\n")
print("=== HAK AKSES PELANGGAN ===")

print("Kode Hak Akses             :", kode_hak_akses_2024)
print("Member Access              :", member_access_2024)
print("Promo Access               :", promo_access_2024)
print("Free Shipping Access       :", free_shipping_2024)

# OUTPUT OPERASI BITWISE
print("\n")
print("=== OPERASI BITWISE ===")

print("=== Kode Status Transaksi ===")

print("0001 | 0010 | 0100 | 1000")

print("Kode Biner   :", format(kode_status_2024, "04b"))
print("Kode Desimal :", kode_status_2024)

# PEMERIKSAAN STATUS DENGAN AND
print("\n=== Pemeriksaan Status ===")

print("\nCek Member")
print(format(kode_status_2024, "04b"), "&", format(bit_member_2024, "04b"))

print("Hasil Biner   :", format(cek_member_2024, "04b"))
print("Hasil Desimal :", cek_member_2024)


print("\nCek Promo")
print(format(kode_status_2024, "04b"), "&", format(bit_promo_2024, "04b"))

print("Hasil Biner   :", format(cek_promo_2024, "04b"))
print("Hasil Desimal :", cek_promo_2024)

# PERBANDINGAN STATUS DENGAN XOR
print("\n=== Perbandingan Status ===")

print("Kode Transaksi :", format(kode_status_2024, "04b"))
print("Kode Referensi :", format(kode_referensi_2024, "04b"))

print(
    format(kode_status_2024, "04b"),
    "^",
    format(kode_referensi_2024, "04b")
)

print("Hasil Biner   :", format(perbandingan_status_2024, "04b"))
print("Hasil Desimal :", perbandingan_status_2024)

# SHIFT
print("\n=== Shift ===")

print(format(kode_status_2024, "04b"), "<< 1")

print("Hasil Biner   :", format(hasil_shift_2024, "b"))
print("Hasil Desimal :", hasil_shift_2024)

# OUTPUT OPERATOR IDENTITAS
print("\n=== OPERATOR IDENTITAS ===")

print("kode_1 is kode_3     :", identitas_sama_2024)
print("kode_1 is not kode_2 :", identitas_berbeda_2024)
print("kode_1 == kode_2     :", nilai_sama_2024)

# SELESAI
print("\n=== SELESAI ===")
````

###Jawaban:

1. Operator aritmatika yang digunakan dalam program adalah perkalian (*), pengurangan (-), pembagian (/), dan modulus (%). Operator perkalian digunakan untuk menghitung besar diskon berdasarkan total belanja dan persentase diskon. Operator pengurangan digunakan untuk memperoleh total pembayaran setelah diskon dikurangi dari total belanja. Operator pembagian digunakan untuk menghitung rata-rata harga setiap barang, sedangkan operator modulus digunakan untuk mengetahui sisa dari total belanja ketika dibagi dengan 1000.
2. Operator and digunakan ketika dua kondisi harus terpenuhi secara bersamaan. Pada program, diskon_member_2024 bernilai benar apabila pelanggan merupakan member dan total belanjanya minimal Rp200.000. Operator or digunakan ketika cukup salah satu kondisi yang terpenuhi. Pada mendapatkan_promo_2024, pelanggan mendapatkan promo apabila memenuhi syarat diskon member atau memenuhi syarat jumlah barang dan promo tersedia. Sementara itu, operator not digunakan untuk membalik nilai kondisi. Pada bukan_member_2024, nilai menjadi benar apabila pelanggan bukan merupakan member.
3. Operator in digunakan untuk memeriksa apakah kode promo yang dimasukkan pelanggan terdapat di dalam daftar_promo_2024. Jika kode tersebut tersedia, promo_tersedia_2024 bernilai True. Sebaliknya, operator not in digunakan untuk memeriksa apakah kode promo tidak terdapat dalam daftar tersebut. Hasil pemeriksaan kemudian dapat digunakan untuk menentukan apakah pelanggan memiliki akses terhadap promo.
4. Operator is digunakan untuk memeriksa apakah dua variabel menunjuk pada objek yang sama, sedangkan == digunakan untuk membandingkan apakah nilai dari dua objek sama. Pada program, kode_3_2024 dibuat dengan menunjuk objek yang sama dengan kode_1_2024, sehingga pemeriksaan kode_1_2024 is kode_3_2024 digunakan untuk melihat kesamaan identitas objek. Sementara itu, kode_1_2024 == kode_2_2024 digunakan untuk membandingkan nilai kedua list. Walaupun kode_1_2024 dan kode_2_2024 memiliki isi yang sama, keduanya dibuat sebagai objek yang berbeda.
5. Operator bitwise digunakan untuk menyimpan beberapa kondisi pelanggan dalam bentuk bit. Program menggunakan 0001 untuk status member, 0010 untuk memenuhi minimum belanja, 0100 untuk jumlah barang minimal tiga, dan 1000 untuk kode promo yang tersedia. Operator OR (|) digunakan untuk menggabungkan kondisi-kondisi tersebut menjadi kode_status_2024. Selanjutnya, operator AND (&) digunakan untuk memeriksa apakah status tertentu terdapat dalam kode tersebut. Operator XOR (^) digunakan untuk membandingkan kode status transaksi dengan kode referensi, sedangkan shift kiri (<<) digunakan untuk menggeser posisi bit.
6. Program dimulai dengan meminta input berupa nama pelanggan, status pelanggan, total belanja, jumlah barang, dan kode promo. Setelah itu, program menentukan apakah total belanja minimal Rp200.000, jumlah barang minimal tiga, dan pelanggan merupakan member. Kode promo kemudian diperiksa menggunakan operator in dan not in. Hasil pemeriksaan tersebut diproses menggunakan operator logika untuk menentukan apakah pelanggan mendapatkan diskon dan promo. Jika pelanggan merupakan member dan memenuhi minimum belanja, diberikan diskon sebesar 10%. Selanjutnya, diskon dihitung menggunakan perkalian dan dikurangkan dari total belanja untuk memperoleh total pembayaran. Program kemudian menampilkan data transaksi, hasil validasi, perhitungan, hak akses, serta hasil operasi bitwise.
7. Salah satu kondisi yang dapat menghasilkan True adalah belanja_minimum_2024 = belanja_2024 >= 200000 apabila pelanggan memasukkan total belanja sebesar Rp200.000 atau lebih. Contoh kondisi yang dapat menghasilkan False adalah status_member_2024 = status_2024 == "member" apabila pelanggan memasukkan status nonmember. Dengan demikian, hasil operator perbandingan bergantung pada data yang dimasukkan oleh pengguna.
