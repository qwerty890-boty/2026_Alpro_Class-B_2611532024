## 2.1 Dasar Teori

### 2.1.1 Python

Python merupakan bahasa pemrograman tingkat tinggi yang memiliki sintaks sederhana dan mudah dibaca. Python dapat digunakan untuk berbagai kebutuhan, mulai dari pembelajaran pemrograman hingga pengembangan perangkat lunak.

Pada tahap awal pembelajaran Python, salah satu perintah dasar yang dipelajari adalah `print()`. Perintah tersebut digunakan untuk menampilkan teks atau informasi ke layar.

Contoh:

```python
print("Hello World")
```

### 2.1.2 Perintah print()

Fungsi print() digunakan untuk menampilkan suatu informasi atau teks ke layar. Teks pada Python umumnya ditulis di antara tanda kutip.

Contoh:
```python
print("Hello World")
print("Belajar Python")
```
Hasilnya:
```python
Hello World
Belajar Python
```

### 2.1.3 Komentar

Komentar merupakan bagian dari kode yang tidak akan dijalankan oleh Python. Komentar digunakan untuk memberikan keterangan atau penjelasan mengenai kode program.

Komentar satu baris dapat dibuat menggunakan tanda `#`.

Contoh:
```python
# Program pertama Python
print("Hello World")
```
Python juga dapat menggunakan tanda kutip tiga (""") untuk membuat teks yang terdiri dari beberapa baris.

###2.1.4 Escape Character

Escape character digunakan untuk memberikan fungsi khusus pada karakter tertentu di dalam string.

Contohnya adalah `\n` yang digunakan untuk membuat baris baru.

Contoh:
```python
print("Baris pertama\nBaris kedua")
```
Hasilnya:
```python
Baris pertama
Baris kedua
```
###2.1.5 Penggabungan Teks

Python dapat digunakan untuk menggabungkan beberapa teks menggunakan operator `+`.

Contoh:
```python
print("Program " + "Python")
```
Hasilnya:
```python
Program Python
```
###2.1.6 Visual Studio Code

Visual Studio Code atau VS Code merupakan editor kode yang dapat digunakan untuk menulis dan menjalankan berbagai bahasa pemrograman, termasuk Python.

Dalam praktikum, VS Code digunakan untuk membuat file Python dengan ekstensi .py, menulis kode, dan menjalankan program.

###2.1.7 GitHub

GitHub merupakan platform yang dapat digunakan untuk menyimpan dan mengelola kode program secara daring. GitHub dapat digunakan bersama Git untuk mencatat perubahan kode dan menyimpan proyek dalam sebuah repository.

## 2.3 Tahapan Commit dan Push ke GitHub

Setelah file program dan laporan selesai dibuat, kedua file dapat dimasukkan ke repository GitHub menggunakan Git.

**1. Inisialisasi repository Git:**

```bash
git init
```

**2. Menambahkan file ke staging area:**

```bash
git add tugas1_2611532024.py LapPrak_ALpro1.ipynb
```

**3. Membuat commit:**

```bash
git commit -m "Tugas 1 dan laporan praktikum pekan 1"
```

**4. Menghubungkan repository lokal dengan repository GitHub:**

```bash
git remote add origin https://github.com/USERNAME/NAMA-REPOSITORY.git
```

**5. Menggunakan branch `main`:**

```bash
git branch -M main
```

**6. Mengirim file ke GitHub:**

```bash
git push -u origin main
```

Setelah proses push berhasil, file `tugas1_2611532024.py` dan `LapPrak_ALpro1.ipynb` akan tersimpan di repository GitHub.
