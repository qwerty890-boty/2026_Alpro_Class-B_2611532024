## 2.1.1 Perulangan pada Python

Perulangan atau *looping* merupakan salah satu struktur kontrol dalam pemrograman yang digunakan untuk menjalankan suatu instruksi secara berulang. Perulangan sangat berguna ketika suatu proses harus dilakukan beberapa kali, sehingga programmer tidak perlu menuliskan perintah yang sama secara berulang-ulang.

Python menyediakan beberapa jenis perulangan, salah satunya adalah `for`. Perulangan `for` digunakan untuk melakukan iterasi terhadap suatu objek yang dapat diiterasi atau disebut **iterable**.

**Iterable** adalah objek dalam Python yang elemennya dapat diakses atau dilalui satu per satu. Beberapa contoh iterable adalah `list`, `tuple`, `string`, `set`, `dictionary`, dan `range`.

Contoh:

```python
buah = ["Apel", "Mangga", "Jeruk"]

for item in buah:
    print(item).
```
Pada program tersebut, `buah` merupakan iterable karena memiliki beberapa elemen yang dapat diambil satu per satu. Variabel `item` akan menerima nilai setiap elemen secara berurutan.

##2.1.2 Perulangan for

Perulangan `for` merupakan struktur perulangan yang digunakan untuk mengulang suatu blok kode berdasarkan elemen-elemen yang terdapat dalam sebuah iterable.

Bentuk umum perulangan `for` adalah:
```python
for variabel in iterable:
    perintah
```
Hasil:
```
1
2
3
4
5
```
Pada contoh tersebut, `range(1, 6)` menghasilkan bilangan dari 1 sampai 5. Variabel `i` akan menerima setiap nilai tersebut secara berurutan, kemudian `print(i)` dijalankan pada setiap iterasi.

Perulangan `for` juga dapat digunakan untuk mengakses setiap karakter dalam sebuah string.

Contoh:
```python
kata = "Python"

for huruf in kata:
    print(huruf)
```
Hasil:
```
P
y
t
h
o
n
```
Dari contoh tersebut dapat diketahui bahwa string `"Python"` merupakan iterable karena setiap karakter di dalamnya dapat diakses satu per satu.

##2.1.3 Nested for

Nested for adalah perulangan `for` yang berada di dalam perulangan `for` lainnya. Perulangan yang berada di bagian luar disebut **outer loop**, sedangkan perulangan yang berada di dalamnya disebut **inner loop**.

Bentuk umum nested for adalah:
```python
for variabel_luar in iterable_luar:
    for variabel_dalam in iterable_dalam:
        perintah
```
Pada nested for, setiap satu kali iterasi pada outer loop akan menyebabkan seluruh proses pada inner loop dijalankan sampai selesai.

Contoh:
```pyhton
for i in range(1, 4):
    for j in range(1, 4):
        print(i, j)
```
Hasil:
```
1 1
1 2
1 3
2 1
2 2
2 3
3 1
3 2
3 3
```

Pada program tersebut, ketika `i` bernilai 1, inner loop menjalankan seluruh nilai `j`, yaitu 1, 2, dan 3. Setelah inner loop selesai, nilai `i` berubah menjadi 2 dan inner loop kembali dijalankan dari awal. Proses tersebut berlanjut hingga seluruh iterasi outer loop selesai.

Nested for dapat digunakan untuk menyelesaikan permasalahan yang memiliki struktur bertingkat, seperti mencetak pola, mengolah data berbentuk tabel, dan melakukan operasi pada baris dan kolom matriks.

Contoh penggunaan nested `for` untuk mencetak pola:
```python
for i in range(1, 5):
    for j in range(i):
        print("*", end="")
    print()
```
Hasil:
```
*
**
***
****
```
Pada contoh tersebut, outer loop digunakan untuk menentukan jumlah baris, sedangkan inner loop digunakan untuk menentukan jumlah karakter * yang dicetak pada setiap baris.

### 2.2.3 `perulangan_for3_2611532024.py`

````python
#Buat file dengan nama perulangan_for3_2611532024.py
#Buat program untuk perulangan for dalam Python
#Nama variabel ditambah 4 digit nim terakhir contoh: ulang_1234
#Program ini menggunakan fungsi input()

ulang_2024 = int(input("Masukkan jumlah perulangan: "))

jumlah_2024 = 0
for i in range(1, ulang_2024 + 1):
    print(i, end="")
    jumlah_2024 = jumlah_2024 + i

    if i < ulang_2024:
        print(" + ", end="")
    else:
        print(" = ", jumlah_2024, end=" ")
print()
print("jumlah =", jumlah_2024)
````

**Penjelasan Program**

Program ini menggunakan perulangan `for` untuk menghitung jumlah dari angka `1` sampai batas perulangan yang dimasukkan pengguna. Variabel `jumlah_2024` digunakan untuk menyimpan hasil penjumlahan setiap angka.

Pada setiap perulangan, nilai `i` ditampilkan dan kemudian ditambahkan ke `jumlah_2024`. Kondisi `if` digunakan untuk memberikan tanda `+` di antara angka dan tanda `=` sebelum hasil akhir.

Setelah perulangan selesai, program menampilkan hasil penjumlahan dengan format `jumlah =`.

### 2.2.4 `jumlah_genap_2611532024.py`

### 2.2.5 `nested_for1_2611532024.py`

````python
#Buat file dengan nama nested_for1_2611532024.py
#Buat program untuk perulangan for dalam Python
#Nama variabel ditambah 4 digit nim terakhir contoh: ulang_1234
#Program ini menggunakan fungsi input()

batas_2024 = int(input("Masukkan nilai batas: "))
for line in range(1, batas_2024 + 1):
    for j in range(1,(-1* line + batas_2024)+1):
        print(".", end= " ")
    print(line)
````

**Penjelasan Program**

Program ini menggunakan **nested `for`** atau perulangan `for` di dalam `for` untuk menampilkan pola titik dan angka. Perulangan pertama menggunakan variabel `line` untuk menentukan baris, sedangkan perulangan kedua menggunakan `j` untuk menentukan jumlah titik pada setiap baris.

Jumlah titik dihitung berdasarkan nilai `(-1 * line + batas_2024)`, sehingga jumlah titik akan berkurang setiap baris. Setelah perulangan kedua selesai, nilai `line` ditampilkan pada akhir baris.

###2.2.8 `nested_for4_2611532024.py`

````python
# Buat file dengan nama nested_for4_2611532024.py
# Buat program untuk perulangan for dalam Python
# Nama variabel ditambah 4 digit nim terakhir
# Program ini menggunakan fungsi input()

tinggi_2024 = int(input("Masukkan tinggi pola (bilangan genap, misal 10: )"))

if tinggi_2024 % 2 != 0:
    print("Tinggi harus bilangan genap!")
else:
    a_2024 = tinggi_2024
    c_2024 = a_2024
    lebar_2024 = (2 * tinggi_2024) - 2

    for i_2024 in range(1, tinggi_2024 + 1):
        b_2024 = c_2024 + 1

        for j_2024 in range(1, lebar_2024 + 1):

            # Baris atas dan bawah
            if i_2024 == 1 or i_2024 == tinggi_2024:
                if j_2024 == 1 or j_2024 == lebar_2024:
                    print("#", end="")
                else:
                    print("=", end="")

            # Baris isi
            else:
                if j_2024 == 1 or j_2024 == lebar_2024:
                    print("|", end="")
                else:
                    if j_2024 == c_2024:
                        print("<", end="")
                    elif j_2024 == b_2024:
                        print(">", end="")
                    elif j_2024 == (lebar_2024 - c_2024):
                        print("<", end="")
                    elif j_2024 == (lebar_2024 - c_2024 + 1):
                        print(">", end="")
                    elif j_2024 > b_2024 and j_2024 < (lebar_2024 - c_2024):
                        print(".", end="")
                    else:
                        print(" ", end="")

        print()

        # Logika asli Java
        a_2024 -= 2

        if a_2024 <= 0:
            c_2024 = (-a_2024) + 2
        else:
            c_2024 = a_2024
````

**Penjelasan Program**

Program ini menggunakan **nested `for`** untuk membuat pola dengan tinggi dan lebar yang ditentukan berdasarkan input pengguna. Program terlebih dahulu memastikan `tinggi_2024` merupakan bilangan genap.

Perulangan pertama menggunakan `i_2024` untuk menentukan baris, sedangkan perulangan kedua menggunakan `j_2024` untuk menentukan posisi karakter pada setiap baris. Kondisi `if` digunakan untuk mengatur karakter `#`, `=`, `|`, `<`, `>`, dan `.` sesuai dengan posisi masing-masing.

Variabel `c_2024` dan `a_2024` digunakan untuk mengatur perubahan posisi pola pada setiap baris, sehingga pola dapat berubah secara bertahap dari bagian atas hingga bagian bawah.

````python
# ============================================================
# PROGRAM JAM PASIR KRISTAL PALINDROMIK
# PEKAN 5
# ============================================================

print("=== PROGRAM JAM PASIR KRISTAL PALINDROMIK ===")

# Input ukuran N
n_2024 = int(input("Masukkan ukuran skala jam pasir (N): "))


# ============================================================
# BORDER ATAS
# ============================================================

print("#", end="")

for angka_2024 in range(4 * n_2024 + 5):
    print("=", end="")

print("#")


# ============================================================
# FASE 1: JAM PASIR ATAS
# Baris N turun sampai 1
# ============================================================

for baris_2024 in range(n_2024, 0, -1):

    # Pembatas kiri
    print("|", end="")
    print(" ", end="")

    # Spasi penyeimbang kiri
    for spasi_2024 in range(2 * (n_2024 - baris_2024)):
        print(" ", end="")

    # Angka menurun
    for angka_2024 in range(baris_2024, 0, -1):
        print(angka_2024, end="")
        print(" ", end="")

    # Poros kristal
    print("<*>", end="")

    # Angka menaik
    for angka_2024 in range(1, baris_2024 + 1):
        print(" ", end="")
        print(angka_2024, end="")

    # Spasi penyeimbang kanan
    for spasi_2024 in range(2 * (n_2024 - baris_2024)):
        print(" ", end="")

    # Pembatas kanan
    print(" ", end="")
    print("|", end="")

    # Pindah baris
    print()


# ============================================================
# FASE 2: POROS TITIK PUSAT
# ============================================================

print("|", end="")

# Spasi kiri
for spasi_2024 in range(2 * n_2024 + 1):
    print(" ", end="")

# Poros kristal
print("<*>", end="")

# Spasi kanan
for spasi_2024 in range(2 * n_2024 + 1):
    print(" ", end="")

# Pembatas kanan
print("|")



# ============================================================
# FASE 3: JAM PASIR BAWAH
# Baris 1 naik sampai N
# ============================================================

for baris_2024 in range(1, n_2024 + 1):

    # Pembatas kiri
    print("|", end="")
    print(" ", end="")

    # Spasi penyeimbang kiri
    for spasi_2024 in range(2 * (n_2024 - baris_2024)):
        print(" ", end="")

    # Angka menurun
    for angka_2024 in range(baris_2024, 0, -1):
        print(angka_2024, end="")
        print(" ", end="")

    # Poros kristal
    print("<*>", end="")

    # Angka menaik
    for angka_2024 in range(1, baris_2024 + 1):
        print(" ", end="")
        print(angka_2024, end="")

    # Spasi penyeimbang kanan
    for spasi_2024 in range(2 * (n_2024 - baris_2024)):
        print(" ", end="")

    # Pembatas kanan
    print(" ", end="")
    print("|", end="")

    # Pindah baris
    print()


# ============================================================
# BORDER BAWAH
# ============================================================

print("#", end="")

for angka_2024 in range(4 * n_2024 + 5):
    print("=", end="")

print("#")
````

### 2.3.2 Penjelasan Program

### Penjelasan Program

Program ini menggunakan **nested `for`** untuk membuat pola **jam pasir kristal palindromik** berdasarkan ukuran `n_2024` yang dimasukkan pengguna.

Program terdiri dari **border atas, fase jam pasir atas, poros titik pusat, fase jam pasir bawah, dan border bawah**. Perulangan `for` digunakan untuk mengatur jumlah baris, spasi, serta urutan angka yang membentuk pola.

Pada setiap baris, angka dicetak secara **menurun** kemudian memiliki poros kristal `<*>`, lalu angka dicetak kembali secara **menaik**. Fase atas menggunakan `range(n_2024, 0, -1)`, sedangkan fase bawah menggunakan `range(1, n_2024 + 1)` sehingga pola bagian bawah menjadi kebalikan dari bagian atas.

# BAB 3 KESIMPULAN

Berdasarkan praktikum yang telah dilakukan, dapat disimpulkan bahwa perulangan merupakan salah satu konsep penting dalam pemrograman Python yang digunakan untuk menjalankan suatu instruksi secara berulang sehingga program menjadi lebih efisien dan terstruktur.

Perulangan `for` digunakan untuk melakukan iterasi terhadap suatu iterable, yaitu objek yang elemennya dapat diakses atau dilalui satu per satu, seperti `list`, `tuple`, `string`, `set`, `dictionary`, dan `range`. Dengan menggunakan for, programmer dapat menjalankan suatu perintah untuk setiap elemen tanpa harus menuliskan perintah yang sama secara berulang.

Selain itu, nested for merupakan perulangan `for` yang ditempatkan di dalam perulangan `for` lainnya. Nested for memungkinkan program melakukan proses pengulangan secara bertingkat dan dapat diterapkan dalam berbagai permasalahan, seperti pembuatan pola, pengolahan tabel, dan pemrosesan matriks.

Dengan memahami konsep perulangan `for` dan nested for, mahasiswa dapat membuat program yang lebih sederhana, terstruktur, dan efisien dalam menyelesaikan permasalahan yang membutuhkan proses pengulangan.
