# BAB 2 PEMBAHASAN 2.1 Dasar Teori

## 2.1.1 `if`

`if` merupakan struktur percabangan paling sederhana dalam Python. Struktur ini digunakan ketika program hanya perlu menjalankan suatu blok kode apabila sebuah kondisi terpenuhi.

Bentuk umum penggunaan `if` adalah:

```python
if kondisi:
    pernyataan
```
Python akan mengevaluasi kondisi yang diberikan. Apabila kondisi bernilai True, pernyataan yang berada di dalam blok if akan dijalankan. Sebaliknya, apabila kondisi bernilai False, blok tersebut akan dilewati dan program melanjutkan instruksi berikutnya.
Contohnya:
```pyhton
umur_2024 = 20

if umur_2024 >= 17:
    print("Sudah memenuhi batas usia.")
```
Pada contoh tersebut, pesan akan ditampilkan apabila `nilai umur_2024` lebih besar atau sama dengan 17.

## 2.1.2 `if-else`

`if-else` digunakan ketika terdapat dua kemungkinan keputusan. Jika kondisi pada `if` bernilai `True`, program menjalankan blok `if`. Jika kondisi bernilai `False`, program menjalankan blok `else`.

Bentuk umum penggunaan `if-else` adalah:
```pyhton
if kondisi:
    pernyataan_jika_benar
else:
    pernyataan_jika_salah
```
Contohnya:
```pyhton
nilai_2024 = 85

if nilai_2024 >= 90:
    print("A")
elif nilai_2024 >= 80:
    print("B")
elif nilai_2024 >= 70:
    print("C")
else:
    print("D")
```
Pada contoh tersebut, nilai `75` memenuhi kondisi `nilai_2024 >= 60`, sehingga program menampilkan pesan "Lulus". Jika kondisi tersebut tidak terpenuhi, program akan menjalankan bagian `else`.

## 2.1.3 `if-elif-else`

`if-elif-else` digunakan ketika program memiliki lebih dari dua kemungkinan kondisi. Python akan memeriksa kondisi secara berurutan dari atas ke bawah. Ketika menemukan kondisi yang bernilai `True`, blok tersebut akan dijalankan dan pemeriksaan pada kondisi berikutnya tidak dilanjutkan.

Bentuk umum penggunaan `if-elif-else` adalah:
```pyhton
if kondisi_1:
    pernyataan_1
elif kondisi_2:
    pernyataan_2
else:
    pernyataan_lain
```
Contohnya:
```pyhton
nilai_2024 = 85

if nilai_2024 >= 90:
    print("A")
elif nilai_2024 >= 80:
    print("B")
elif nilai_2024 >= 70:
    print("C")
else:
    print("D")
```
Pada contoh tersebut, program memeriksa kondisi secara berurutan. Karena nilai 85 tidak memenuhi kondisi pertama tetapi memenuhi kondisi kedua, program menampilkan "B".

Struktur ini dapat digunakan untuk menentukan kategori atau tingkatan berdasarkan nilai tertentu.

## 2.1.4 `multi-if`

`multi-if` menggunakan beberapa pernyataan `if` yang berdiri sendiri. Berbeda dengan `if-elif-else`, setiap kondisi pada `multi-if` diperiksa secara terpisah. Oleh karena itu, lebih dari satu kondisi dapat bernilai `True` dan menjalankan blok programnya masing-masing.
Contohnya:
```pyhton
belanja_2024 = 250000

if belanja_2024 >= 100000:
    print("Mendapat diskon pertama")

if belanja_2024 >= 200000:
    print("Mendapat diskon kedua")
```
Pada contoh tersebut, kedua kondisi bernilai `True`, sehingga kedua pesan akan ditampilkan.

Struktur `multi-if` dapat digunakan ketika beberapa kondisi dapat terpenuhi secara bersamaan. Salah satu penerapannya adalah menghitung diskon secara akumulatif berdasarkan beberapa ketentuan yang berbeda.

## 2.1.5 `match-case`

`match-case` digunakan untuk mencocokkan suatu nilai dengan beberapa kemungkinan pilihan. Struktur ini tersedia pada Python versi 3.10 dan yang lebih baru.

Bentuk umum penggunaan `match-case` adalah:
```pyhton
match nilai:
    case pilihan_1:
        pernyataan_1
    case pilihan_2:
        pernyataan_2
    case _:
        pernyataan_default
```
Contohnya:
```pyhton
pilihan_2024 = 2

match pilihan_2024:
    case 1:
        print("Wahana A")
    case 2:
        print("Wahana B")
    case 3:
        print("Wahana C")
    case _:
        print("Pilihan tidak tersedia")
```
Pada contoh tersebut, `nilai pilihan_2024` akan dicocokkan dengan setiap `case`. Karena nilainya adalah 2, program menjalankan blok `case 2`.

`case _` digunakan sebagai pola umum atau kondisi default ketika nilai yang diberikan tidak sesuai dengan pilihan case yang tersedia.

### 2.2.3 `if_elif_else1_2611532024.py`

````python
#Buat file dengan nama if_elif_else1_2611532024.py
#Buat program untuk kondisional if
#Nama variabel ditambah 4 digit nim terakhir contoh: ipk_1234
#Program ini menggunakan fungsi input()

umur_2024 = int(input("Input umur anda: "))
sim_2024 = input("Apakah Anda Sudah Punya Sim C (y/t:) ")[0]

if umur_2024 >= 17 and sim_2024 == 'y':
    print("Anda Sudah dewasa dan Boleh bawa motor")
elif umur_2024 >= 17 and sim_2024 != 'y':
    print("Anda Sudah dewasa tetapi tidak boleh bawa motor")
elif umur_2024 < 17 and sim_2024 == 'y':
    print("Anda Belum Cukup Umur punya SIM")
else:
    print("Anda Belum Cukup Umur dan tidak boleh bawa motor")
print("Program Selesai")
````

**Penjelasan Program**

Program ini menggunakan struktur percabangan `if-elif-else` untuk menentukan status pengguna berdasarkan umur dan kepemilikan SIM C. Nilai umur disimpan dalam variabel `umur_2024`, sedangkan jawaban kepemilikan SIM C disimpan dalam variabel `sim_2024`.

Program menggunakan operator logika `and` untuk menggabungkan dua kondisi dan operator `!=` untuk memeriksa bahwa nilai `sim_2024` bukan `'y'`.

Program kemudian memeriksa beberapa kondisi secara berurutan, yaitu umur minimal 17 tahun dengan atau tanpa SIM C, serta umur di bawah 17 tahun dengan atau tanpa SIM C. Jika tidak ada kondisi sebelumnya yang terpenuhi, bagian `else` dijalankan. Setelah percabangan selesai, program menampilkan `"Program Selesai"`.

### 2.2.4 `match_case1_2611532024.py`

````python
#Buat file dengan nama match_case1_2611532024.py
#Buat program untuk match case
#Nama variabel ditambah 4 digit nim terakhir contoh: ipk_1234
#Program ini menggunakan fungsi input()
#Program konversi angka menjadi nama bulan

bulan_2024 = int(input("Masukkan angka bulan (1 - 12): "))

match bulan_2024:
    case 1:
        print("Januari")
    case 2:
        print("Februari")
    case 3:
        print("Maret")
    case 4:
        print("April")
    case 5:
            print("Mei")
    case 6:
            print("Juni")
    case 7:
            print("Juli")
    case 8:
            print("Agustus")
    case 9:
            print("September")
    case 10:
            print("Oktober")
    case 11:
            print("November")
    case 12:
            print("Desember")
    case _:
            print("Angka tidak valid")
````

**Penjelasan Program**

Program ini menggunakan struktur `match-case` untuk mengonversi angka yang dimasukkan pengguna menjadi nama bulan. Nilai input disimpan dalam variabel `bulan_2024` dan dikonversi menjadi tipe `int`.

Setiap `case` mencocokkan angka 1 sampai 12 dengan nama bulan yang sesuai. Jika angka yang dimasukkan tidak berada dalam rentang 1–12, maka `case _` dijalankan dan program menampilkan pesan `"Angka tidak valid"`.

### 2.2.5 `multi_if1_2611532024.py`

````python
#Buat file dengan nama multi_if1_2611532024.py
#Buat program untuk kondisional if
#Nama variabel ditambah 4 digit nim terakhir contoh: ipk_1234
#Program ini menggunakan fungsi input()

umur_2024 = int(input("Input umur anda: "))
sim_2024 = input("Apakah Anda Sudah Punya Sim C (y/t:) ")[0]

if umur_2024 >= 17 and sim_2024 == 'y':
    print("Anda Sudah Dewasa dan Boleh bawa motor")

if umur_2024 >= 17 and sim_2024 != 'y':
    print("Anda Sudah Dewasa tetapi tidak boleh bawa motor")

if umur_2024 < 17 and sim_2024 == 'y':
    print("Anda Belum Cukup Umur punya SIM")

if umur_2024 < 17 and sim_2024 != 'y':
    print("Anda Belum Cukup Umur bawa motor")
````

**Penjelasan Program**

Program ini menggunakan konsep `multi-if` untuk memeriksa beberapa kondisi secara terpisah berdasarkan umur dan kepemilikan SIM C. Nilai umur disimpan dalam variabel `umur_2024`, sedangkan jawaban kepemilikan SIM C disimpan dalam variabel `sim_2024`.

Setiap kondisi `if` diperiksa secara independen menggunakan operator logika `and`. Jika kondisi terpenuhi, program akan menampilkan pesan yang sesuai. Karena menggunakan beberapa `if` yang terpisah, setiap kondisi diperiksa tanpa bergantung pada kondisi sebelumnya.

### 2.2.6 `multi_if2_2611532024.py`

````python
#Buat file dengan nama multi_if2_2611532024.py
#Buat program untuk kondisional if
#Nama variabel ditambah 4 digit nim terakhir contoh: total_belanja_1234
#Program ini menggunakan fungsi input()
#Program Menghitung Diskon Belanja

#Input dari user
total_belanja_2024 = float(input("Masukkan total belanja (Rp): "))

#Input status member (mengecek apakah user mengetik 'y' atau 'ya')
input_member_2024 = input("Apakah Anda Member? (y/t )").strip().lower()
is_member_2024 = input_member_2024 in ["y", "ya"]

#Input status kode promo (mengecek apakah user mengetik 'y' atau 'ya')
input_promo_2024 = input("Apakah kode promo valid? (y/t): ").strip().lower()
kode_promo_valid_2024 = input_promo_2024 in ["y", "ya"]

total_diskon_persen_2024 = 0

#Multi if terpisah: Setiap kondisi diperiksa secara independen
#Diskon bisa ditumpuk (akumulasi) jika memenuhi beberapa syarat sekaligus

if total_belanja_2024 > 1000000:
    total_diskon_persen_2024 += 10  #Diskon belanja besar

if is_member_2024:
    total_diskon_persen_2024 += 5   #Diskon member

if kode_promo_valid_2024:
    total_diskon_persen_2024 += 15 #Diskon voucher

#Menghitung nominal diskon dan total bayar
nominal_diskon_2024 = total_belanja_2024 * (total_diskon_persen_2024 / 100)
total_bayar_2024 = total_belanja_2024 - nominal_diskon_2024

#Output hasil
print("\n--- Rincian Pembayaran ---")
print(f"Total Diskon : {total_diskon_persen_2024}% (Rp {nominal_diskon_2024:,.0f})")
print(f"Total Bayar : Rp {total_bayar_2024:,.0f}")

print(f"Total diskon yang Anda dapatkan: {total_diskon_persen_2024}%")
#Output: Total diskon yang Anda dapatkan: 30% jika belanja > 1 juta, member, dan kode promo valid
````

**Penjelasan Program**

Program ini menggunakan konsep `multi-if` untuk menghitung diskon belanja secara akumulatif. Program menerima input total belanja, status member, dan status kode promo dari pengguna.

Status member dan kode promo diperiksa menggunakan operator `in` setelah input diubah menjadi huruf kecil dengan `lower()`. Setiap kondisi `if` diperiksa secara independen. Jika syarat terpenuhi, nilai diskon ditambahkan ke `total_diskon_persen_2024`.

Setelah seluruh kondisi diperiksa, program menghitung nominal diskon dan total pembayaran. Hasilnya kemudian ditampilkan dalam bentuk rincian pembayaran. Jika ketiga kondisi terpenuhi, yaitu belanja lebih dari Rp1.000.000, merupakan member, dan memiliki kode promo yang valid, total diskon yang diperoleh adalah 30%.

Dengan demikian, program menunjukkan penggunaan `if-elif-else` untuk menentukan satu hasil berdasarkan rentang nilai yang diberikan.

### 2.3.1 `tugas4_2611532024.py`

````python
# ============================================================
# SISTEM LOKET ALPRO ADVENTURE PARK
# Tugas 4 Algoritma dan Pemrograman
# ============================================================

print("=== SISTEM LOKET ALPRO ADVENTURE PARK ===")

# ------------------------------------------------------------
# 1. INPUT DATA PENGUNJUNG
# ------------------------------------------------------------

nama_2024 = input("Masukkan Nama Pengunjung        : ")
umur_2024 = int(input("Input umur anda                 : "))
sim_2024 = input("Apakah Anda Sudah Punya SIM C (y/t): ")[0].lower()
jumlah_tiket_2024 = int(input("Masukkan jumlah tiket           : "))

# ------------------------------------------------------------
# 2. IF TUNGGAL
# Validasi jumlah tik
# ------------------------------------------------------------

if jumlah_tiket_2024 <= 0:
    print("Peringatan: kuota tiket tidak valid.")
    print("Program Selesai")
    exit()

# ------------------------------------------------------------
# 3. PEMILIHAN WAHANA MENGGUNAKAN MATCH-CASE
# ------------------------------------------------------------

print("\nPilihan Paket Wahana (1-5):")
print(" 1. Safari Rimba         (Rp 50,000)")
print(" 2. Arung Jeram          (Rp 75,000)")
print(" 3. Motor ATV Ekstrim    (Rp 120,000)")
print(" 4. Roller Coaster Kilat (Rp 100,000)")
print(" 5. All-Access VIP       (Rp 220,000)")

paket_2024 = int(input("Masukkan nomor paket (1-5)      : "))

match paket_2024:
    case 1:
        nama_wahana_2024 = "Wahana Safari Rimba"
        harga_satuan_2024 = 50000

    case 2:
        nama_wahana_2024 = "Wahana Arung Jeram"
        harga_satuan_2024 = 75000

    case 3:
        nama_wahana_2024 = "Wahana Motor ATV Ekstrim"
        harga_satuan_2024 = 120000

    case 4:
        nama_wahana_2024 = "Wahana Roller Coaster Kilat"
        harga_satuan_2024 = 100000

    case 5:
        nama_wahana_2024 = "Wahana All-Access VIP"
        harga_satuan_2024 = 220000

    case _:
        print("Paket wahana tidak valid!")
        print("Program Selesai")
        exit()

# ------------------------------------------------------------
# Input data tambahan
# ------------------------------------------------------------

is_member_2024 = input("Apakah Anda member? (y/t)       : ").strip().lower()
kode_promo_valid_2024 = input(
    "Apakah kode promo valid? (y/t)  : "
).strip().lower()

# ------------------------------------------------------------
# 4. VALIDASI IZIN KENDALI WAHANA
# Paket 3 menggunakan if-elif-else dengan AND dan !=
# Paket lainnya menggunakan if-else
# ------------------------------------------------------------

print("\n--- KELAYAKAN PENGENDARA WAHANA ---")

if paket_2024 == 3:
    if umur_2024 >= 17 and sim_2024 == 'y':
        print("Status Akses: Anda sudah dewasa dan boleh mengendarai ATV sendiri.")

    elif umur_2024 >= 17 and sim_2024 != 'y':
        print(
            "Status Akses: Anda sudah dewasa tetapi tidak boleh "
            "bawa motor ATV (wajib didampingi instruktur)."
        )

    elif umur_2024 < 17 and sim_2024 == 'y':
        print(
            "Status Akses: Identitas tidak valid: "
            "Belum cukup umur memiliki SIM."
        )

    else:
        print(
            "Status Akses: Anda belum cukup umur dan "
            "tidak boleh bawa motor ATV."
        )

else:
    if umur_2024 >= 10:
        print("Status Akses: Anda memenuhi batas usia wahana.")
    else:
        print("Status Akses: Anda belum memenuhi batas usia wahana.")

# ------------------------------------------------------------
# 5. PERHITUNGAN SUBTOTAL
# ------------------------------------------------------------

subtotal_2024 = harga_satuan_2024 * jumlah_tiket_2024

# ------------------------------------------------------------
# 6. AKUMULASI DISKON MENGGUNAKAN MULTI-IF
# Semua IF dibuat TERPISAH agar diskon dapat ditumpuk.
# ------------------------------------------------------------

total_diskon_persen_2024 = 0

# Diskon Belanja Besar
if subtotal_2024 >= 200000:
    total_diskon_persen_2024 += 10

# Diskon Member
if is_member_2024 in ['y', 'ya']:
    total_diskon_persen_2024 += 5

# Diskon Voucher Promo
if kode_promo_valid_2024 in ['y', 'ya']:
    total_diskon_persen_2024 += 15

# Diskon Tambahan Rombongan
if jumlah_tiket_2024 >= 5:
    total_diskon_persen_2024 += 5

# ------------------------------------------------------------
# 7. PERHITUNGAN PEMBAYARAN
# ------------------------------------------------------------

nominal_diskon_2024 = subtotal_2024 * (
    total_diskon_persen_2024 / 100
)

total_bayar_2024 = subtotal_2024 - nominal_diskon_2024

# ------------------------------------------------------------
# 8. EVALUASI KELULUSAN AUDIT MENGGUNAKAN IF-ELSE
# ------------------------------------------------------------

if total_bayar_2024 > 300000:
    catatan_layanan_2024 = (
        "Selamat! Anda berhak mendapatkan Souvenir Gratis."
    )
else:
    catatan_layanan_2024 = "Terima kasih telah berkunjung."

# ------------------------------------------------------------
# 9. RINCIAN PEMBAYARAN
# ------------------------------------------------------------

print("\n--- Rincian Pembayaran ---")
print(f"Nama Pengunjung  : {nama_2024}")
print(f"Wahana           : {nama_wahana_2024}")
print(f"Jumlah Tiket     : {jumlah_tiket_2024}")
print(f"Subtotal Belanja : Rp {subtotal_2024:,.0f}")
print(
    f"Total Diskon     : {total_diskon_persen_2024}% "
    f"(Rp {nominal_diskon_2024:,.0f})"
)
print(f"Total Bayar      : Rp {total_bayar_2024:,.0f}")
print(f"Catatan Layanan  : {catatan_layanan_2024}")

print("Program Selesai")
````

### 2.3.2 Penjelasan Program

Program `tugas4_2611532024.py` merupakan penerapan beberapa struktur percabangan Python dalam sistem loket Alpro Adventure Park. Program menerima data pengunjung berupa nama, umur, kepemilikan SIM C, dan jumlah tiket.

Pertama, struktur `if` tunggal digunakan untuk memeriksa jumlah tiket. Jika jumlah tiket kurang dari atau sama dengan 0, program menampilkan peringatan dan dihentikan menggunakan `exit()`.

Selanjutnya, `match-case` digunakan untuk menentukan paket wahana berdasarkan nomor yang dipilih. Setiap `case` menentukan nama wahana dan harga tiket. Jika pilihan tidak tersedia, `case _` dijalankan dan program dihentikan.

Pada bagian kelayakan pengendara, program menggunakan kondisi umur dan kepemilikan SIM untuk menentukan status akses. Untuk paket ATV, kondisi menggunakan operator `and` dan `!=` untuk memeriksa umur serta kepemilikan SIM. Untuk paket lainnya, digunakan `if-else` untuk memeriksa batas usia.

Setelah itu, program menghitung subtotal berdasarkan harga satuan dikalikan jumlah tiket. Diskon dihitung menggunakan beberapa `if` terpisah sehingga setiap diskon dapat diakumulasikan apabila persyaratannya terpenuhi. Diskon diberikan berdasarkan jumlah belanja, status member, kode promo, dan jumlah tiket.

Program kemudian menghitung nominal diskon dan total pembayaran. Terakhir, `if-else` digunakan untuk menentukan catatan layanan berdasarkan total pembayaran. Seluruh informasi transaksi kemudian ditampilkan dan program diakhiri dengan pesan `"Program Selesai"`.

# BAB 3 KESIMPULAN

Struktur percabangan merupakan salah satu konsep dasar yang penting dalam pemrograman karena memungkinkan program mengambil keputusan berdasarkan kondisi tertentu. Python menyediakan beberapa bentuk struktur percabangan, yaitu `if`, `if-else`, `if-elif-else`, `multi-if`, dan `match-case`, yang memiliki karakteristik dan penggunaan yang berbeda.

`if` digunakan ketika program hanya perlu memeriksa satu kondisi. `if-else` digunakan ketika terdapat dua kemungkinan keputusan. `if-elif-else` digunakan ketika terdapat beberapa kondisi yang diperiksa secara berurutan. Sementara itu, `multi-if` digunakan ketika beberapa kondisi perlu diperiksa secara independen dan lebih dari satu kondisi dapat terpenuhi. `match-case` digunakan untuk mencocokkan suatu nilai dengan beberapa pilihan yang telah ditentukan.

Melalui praktikum ini, penerapan struktur percabangan dapat dipahami melalui contoh program dan studi kasus. Penggunaan operator logika dalam kondisi percabangan juga menunjukkan bahwa pengambilan keputusan pada program dapat melibatkan kondisi yang lebih kompleks.

Pemahaman terhadap struktur percabangan menjadi dasar penting dalam pembelajaran algoritma dan pemrograman karena konsep pengambilan keputusan banyak digunakan dalam berbagai jenis program. Dengan menguasai struktur percabangan, program dapat dibuat lebih fleksibel dan mampu memberikan keluaran yang sesuai dengan kondisi yang dihadapi.
